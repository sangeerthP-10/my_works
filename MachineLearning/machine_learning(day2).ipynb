Linear Regression using Multiple Variables

```python
import pandas as pd
from sklearn import linear_model
```

```python
df=pd.DataFrame({
    'area':[2600,3000,3200,3600,4000],
    'bedrooms':[3,4,None,3,5],
    'age':[20,14,18,30,8],
    'price':[550000,565000,610000,595000,760000]
})
df
```

```python
import math
median=math.floor(df.bedrooms.median())
median
```

```python
df.bedrooms.fillna(median,inplace=True)
df
```

```python
df
```

```python
reg=linear_model.LinearRegression()
```

```python
reg.fit(df[['area','bedrooms','age']],df.price)
```

```python
reg.predict([[2000,3,20]])
```

Exercise(Hiring data Prediction)

```python
hiring_df=pd.DataFrame({
    'experience':[None,None,'five','two','seven','three','ten','eleven'],
    'test_score':[8,8,6,10,9,7,None,7],
    'interview_score':[9,6,7,10,6,10,7,8],
    'salary':[50000,45000,60000,65000,70000,62000,72000,80000]
})
hiring_df
```

```python
import pandas as pd
import numpy as np



# Define the mapping dictionary
word_to_num_map = {
    'one': 1,
    'two': 2,
    'three': 3,
    'four': 4,
    'five': 5,
    'six':6,
    'nine':9,
    'seven':7,
    'eleven':11,
    'ten': 10
}

# Use the map function to convert the column
hiring_df['experience_numeric'] = hiring_df['experience'].map(word_to_num_map)

# Print the result and data types to verify
print(hiring_df)

```

```python
hiring_df.drop('experience',axis=1,inplace=True)
```

```python
hiring_df
```

```python
test_score_median=hiring_df['test_score'].median()
hiring_df['test_score'].fillna(test_score_median,inplace=True)
hiring_df.fillna(0,inplace=True)
```

```python

model = linear_model.LinearRegression()

model.fit(hiring_df[['test_score','interview_score','experience_numeric']], hiring_df.salary)
```

**Question1: Predict Salary of candidate who has 2 years of experience,9 test_score, 6 interview_score**

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
prediction=model.predict([[9,6,2]])
print(prediction)
```

**Question 2 : Predict Salary of candidate who has 12 years of experience,10 test_score, 10 interview_score**

```python
model.predict([[10,10,12]])
```
