```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
weekdays=[1,2,3,4,5,6,7]
temperature=[50,45,67,32,43,23,12]
```

```python

plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('weather')
plt.plot(weekdays,temperature,color='red',linewidth=5,linestyle='dotted')
```

```python
days=[1,2,3,4,5,6,7]
max_t=np.array([34,35,37,32,45,37,43])
min_t=np.array([21,30,32,24,35,20,33])
avg=[26,32,33,28,39,26,38]
```

```python

plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('weather')
plt.plot(days,max_t,label='max')
plt.plot(days,avg,label='avg')
plt.plot(days,min_t,label='min')
plt.legend(loc='upper right',fontsize='small')
```

```python

plt.xlabel('Days')
plt.ylabel('temperature')
plt.title('weather')
plt.plot(days,max_t,label='max')
plt.plot(days,avg,label='avg')
plt.plot(days,min_t,label='min')
plt.legend(loc='best',fontsize='small',shadow=True)
plt.grid()
```

```python
company=['A','B','C','D']
revenue=[90,136,89,27]
profit=[40,2,34,12]
```

```python
xpos=np.arange(len(company))
xpos
```

```python
plt.xticks(xpos,company)
plt.bar(xpos-0.2,revenue,width=0.4,label='Revenue')
plt.bar(xpos+0.2,profit,width=0.4,label='profit')
plt.xlabel('company')
plt.ylabel('revenue')
plt.title('Tech Stocks')
plt.legend()
```

```python
plt.yticks(xpos,company)
plt.barh(xpos-0.2,revenue,height=0.4,label='Revenue')
plt.barh(xpos+0.2,profit,height=0.4,label='profit')
plt.xlabel('company')
plt.ylabel('revenue')
plt.title('Tech Stocks')
plt.legend()
```
