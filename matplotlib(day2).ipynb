**Histogram**


```python
import matplotlib.pyplot as plt
```

```python
men_sugar = [112, 85, 129, 94, 76, 138, 101, 119, 88, 107, 92, 124, 81, 133, 98, 115, 79, 103, 127, 90]
```

```python
plt.hist(men_sugar)
```

```python
plt.hist(men_sugar,bins=3)
```

```python
plt.hist(men_sugar,bins=5,rwidth=0.95,color='g')
```

```python
women_sugar = [72, 118, 94, 105, 88, 131, 79, 112, 126, 83, 75, 102, 119, 81, 124, 90, 115]
```

```python
plt.hist([men_sugar,women_sugar],bins=[80,100,120,140,160],rwidth=0.95,color=['green','orange'])
```

```python
plt.hist([men_sugar,women_sugar],bins=[80,100,120,140],rwidth=0.95,color=['green','orange'],label=['men','women'])
plt.xlabel('Sugar Range')
plt.ylabel('Total no. of patient')
plt.title('Blood Sugar Analysis')
plt.legend()
```

```python
plt.hist([men_sugar,women_sugar],bins=[80,100,120,140],rwidth=0.95,color=['green','orange'],label=['men','women'],orientation='horizontal')
plt.xlabel('Sugar Range')
plt.ylabel('Total no. of patient')
plt.title('Blood Sugar Analysis')
plt.legend()
```

**Pie Chart**

```python
expense_value=[1000,600,410,750,300]
explense_labels=["Home Rent","Food","Internet","car","Others"]
```

```python
plt.pie(expense_value,labels=explense_labels)
plt.show()
```

```python
plt.pie(expense_value,labels=explense_labels,radius=2)
plt.show()
```

```python
plt.pie(expense_value,labels=explense_labels,radius=1,autopct='%1.2f%%')
plt.show()
```

```python
plt.pie(expense_value,labels=explense_labels,radius=1,autopct='%1.2f%%',explode=[0,0.3,0,0,0])
plt.show()
```

```python
plt.pie(expense_value,labels=explense_labels,radius=1,autopct='%1.2f%%',explode=[0,0.2,0,0,0.1])
plt.show()
```

```python
plt.pie(expense_value,labels=explense_labels,radius=1,autopct='%1.2f%%',explode=[0,0.2,0,0,0.1],startangle=180)
plt.show()
```
